```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})
```

```python
employees
```

```python
departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
departments
```

```python
pd.merge(employees,departments,on = "DeptID") #Inner Join
```

```python
pd.merge(employees, departments, on="DeptID", how="left")
```

```python
pd.merge(employees, departments, on="DeptID", how="right")
```

```python
pd.merge(employees, departments, on="DeptID", how="outer")
```

Concatenating DataFrames


Use pd.concat() to stack datasets either vertically or horizontally.



```python
df1 = pd.DataFrame({"Name": ["Alice", "Bob"]})
```

```python
df1
```

```python
df2 = pd.DataFrame({"Name": ["Charlie", "David"]})
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)
```

```python
df4 = pd.DataFrame({"Name": ["Alice", "Bob"],"score":[423,634]})
```

```python
df4
```

```python
df3 = pd.DataFrame({"Name": ["Alice", "Bob"],"age":[42,34]})
```

```python
df3
```

```python
pd.concat([df4,df3],axis=1)
```

```python
pd.concat([df4,df3],axis=0)
```
