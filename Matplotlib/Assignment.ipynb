```python
import matplotlib.pyplot as plt
```

```python


years = [2010, 2011, 2012, 2013, 2014,
         2015, 2016, 2017, 2018, 2019]

kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
Rohit = [34, 555, 700, 1500, 1800, 2000, 1700, 1111, 1500, 1900]

plt.plot(years, kohli, label="Virat Kohli", linestyle='--',color='green')
plt.plot(years, sehwag, label="Virender Sehwag", linestyle='-.',color='red')
plt.plot(years, Rohit, label="Rohit Sharma", linestyle=':',color='yellow')

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Player Performance")
plt.legend()
plt.show()
```
