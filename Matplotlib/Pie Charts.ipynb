```python
import matplotlib.pyplot as plt



plt.style.use("ggplot")  # Choose any style you like
```

```python
# Data
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]

plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=labels)
plt.show()
```

```python
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(runs, labels=labels, colors=colors)
plt.show()
```

Add Edges and Style Slices with wedgeprops


```python
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '--'}
)
plt.show()
```

Tip: You can Google “matplotlib wedgeprops” for more customization options.

Visit this documentation page for more



```python

explode = [0.1, 0, 0, 0]  # Highlight Sachin's slice
plt.pie(runs, labels=labels, explode=explode, colors=colors,shadow=True) #Add Shadows for a 3D Feel

plt.title("Exploded Pie Example")
plt.show()
```

```python

explode = [0.1, 0, 0, 0]  # Highlight Sachin's slice
plt.pie(runs, labels=labels, explode=explode, colors=colors,shadow=True,startangle=140) # Use startangle to rotate the pie for better alignment.
plt.title("Exploded Pie Example")
plt.show()
```

```python
explode = [0.1, 0, 0, 0]  # Highlight Sachin's slice
plt.pie(runs, labels=labels, explode=explode, colors=colors, shadow=True, startangle=140, autopct='%1.1f%%')  # Show percentages

plt.title("Exploded Pie Example")
plt.show()
```
