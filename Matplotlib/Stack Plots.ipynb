```python
import matplotlib.pyplot as plt
```

```python
acativities = ['Studying','Playing','Watching TV','Sleeping']
time_spent = [3,2,2,5] # hours in day

colour = ['skyblue','lightgreen','gold','lightcoral']

plt.figure(figsize=(6,6))
plt.pie(time_spent,labels=acativities,colors=colour,autopct='%1.1f%%',startangle=90)
plt.title("After School Activities")
plt.axis('equal')  # Equal aspect ratio ensures the pie is drawn as a circle.
plt.show()

```

```python
days = [1,2,3,4,5,6,7] #day of the week
studying = [3,4,3,5,4,3,4]
playing = [2,2,1,1,2,3,2]
watching_tv = [2,1,2,2,2,1,1]
sleeping = [5,5,6,5,6,5,5]

labels = ['Studyint','playing','Watiching TV', 'seeping']
colour = ['skyblue','lightgreen','gold','lightcoral']

plt.figure(figsize=(10,6))
plt.stackplot(days,studying,playing,watching_tv,sleeping,
              labels=labels, colors=colour,alpha=0.8)
plt.legend(loc='upper left')
plt.title('Weekly Activity Tracker')
plt.xlabel('Day')
plt.ylabel('Hours')
plt.grid(True)
plt.show()
```
